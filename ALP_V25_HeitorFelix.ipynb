10 Peça a quantidade de horas trabalhadas e o valor por hora. Se trabalhou até 40 horas, calcule o salário normalmente (horas × valor por hora). Se trabalhou mais de 40 horas, as horas extras devem ter um adicional de 50% (horas extras × valor por hora × 1.5). Crie um algoritmo em Python que realize esse cálculo corretamente.

```python
horas_trabalhadas = float(input("Digite a quantidade de horas trabalhadas: "))
valor_por_hora = float(input("Digite o valor por hora: "))

if horas_trabalhadas <= 40:
 salario = horas_trabalhadas * valor_por_hora
else:
 horas_extras = horas_trabalhadas - 40
 salario = (40 * valor_por_hora) + (horas_extras * valor_por_hora * 1.5)

print(f"Salário total: R$ {salario:.2f}")
```

```python
def multiplos_de_5():
    n = int(input("Digite o número limite: "))
    print(f"Múltiplos de 5 até {n}:")

    i = 0
    while i <= n:
        if i % 5 == 0:
            print(i)
        i += 1

def numeros_pares():
    n = int(input("Digite o número limite: "))
    print(f"Números pares até {n}:")

    i = 0
    while i <= n:
        if i % 2 == 0:
            print(i)
        i += 1

def numeros_primos():
    n = int(input("Digite o número limite: "))
    print(f"Números primos até {n}:")

    i = 2
    while i <= n:
        primo = True
        j = 2

        while j < i:
            if i % j == 0:
                primo = False
                break
            j += 1

        if primo:
            print(i)

        i += 1

def conversor_temperatura():
    print("1 - Celsius para Fahrenheit")
    print("2 - Fahrenheit para Celsius")
    op = int(input("Escolha: "))

    if op == 1:
        c = float(input("Digite a temperatura em Celsius: "))
        f = (c * 9/5) + 32
        print(f"{c}°C = {f:.2f}°F")

    elif op == 2:
        f = float(input("Digite a temperatura em Fahrenheit: "))
        c = (f - 32) * 5/9
        print(f"{f}°F = {c:.2f}°C")

    else:
        print("Opção inválida!")

opcao = -1

while opcao != 0:
    print("\n===== CALCULADORA v25 =====")
    print("1 - Múltiplos de 5")
    print("2 - Números pares")
    print("3 - Números primos")
    print("4 - Conversor de temperatura")
    print("0 - Sair")
    print("----------------------------")

    opcao = int(input("Digite a opção: "))

    if opcao == 1:
        multiplos_de_5()

    elif opcao == 2:
        numeros_pares()

    elif opcao == 3:
        numeros_primos()

    elif opcao == 4:
        conversor_temperatura()

    elif opcao == 0:
        print("Saindo...")

    else:
        print("Opção inválida!")
```
